```python
#Changing the time column to time data type
tire_wear["Time(s)"] = pd.to_timedelta(tire_wear["Time(s)"],unit="s")
```

```python
#Converting the time column to the dataset index
tire_wear = tire_wear.set_index("Time(s)")
```

```python
#EDA Stage: importing the needed time-series libraries
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf,plot_pacf
warnings.filterwarnings("ignore")
```

```python
#Splitting the dataset into train and test only due to the size of the dataset: later do it with the copy dataset
tire_split = int(len(tire_wear_copy1)*0.7)#70% of the dataset
tire_train = tire_wear_copy1.iloc[:tire_split]
tire_test = tire_wear_copy1.iloc[tire_split:]
```

```python
#Scaling the dataset
train_scaled = scaler.fit_transform(tire_train)
test_scaled = scaler.transform(tire_test)

train_scaled_df = pd.DataFrame(train_scaled,columns=tire_train.columns,index=tire_train.index)
test_scaled_df = pd.DataFrame(test_scaled,columns=tire_test.columns,index=tire_test.index)
```

```python
#ADF at the train dataset to check for stationarity
for column in train_scaled_df.columns:
    stat_result = adfuller(train_scaled_df[column])
    print(f"{column}\nADF Statistic: {stat_result[0]:.4f}\np-value: {stat_result[1]:.4f}")

    if stat_result[1] > 0.05:
        print("Series is non-stationary; needs differencing.")
    else:
        print("Series is stationary; needs no differencing.\n")
```

```python
#Using ACF and PACF plots to select the p,q values
for column in train_scaled_df.columns:
    plot_acf(tire_train[column],title=f"{column} Autocorrelation")
    plt.show()
    plot_pacf(tire_train[column],title=f"{column} Partial Autocorrelation")
    plt.show()
```
